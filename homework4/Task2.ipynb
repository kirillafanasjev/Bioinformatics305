```python
from Bio import motifs
from Bio.Seq import Seq
```

Создадим мотив из посл-тей Задания 1:

```python
sites = ["GAGGTAAAC", "TCCGTAAGC", "CAGGTTGGA",
         "ACAGTCAGC", "TAGGTCAGC", "CAGGTCAGC",
         "CAGGTCGAT", "CAGGTCAGC", "CAGGTCAGC",
         "CAGGTTGGC"]

instances = [Seq(i) for i in sites]
m = motifs.create(instances)
```

```python
pwm = m.counts.normalize(pseudocounts=0.7).log_odds() #возьмем \alpha = 0.7 для избежания беск. значений
```

```python
from Bio import SeqIO

rec = next(SeqIO.parse("chr1.fasta", "fasta"))
sequence_gen = rec.seq
sequence = sequence_gen[:1000000]
```

```python
# Поиск на прямой цепи
threshold = 5.0
hits = []

for position, score in pwm.search(sequence, threshold=threshold):
    hits.append((position, '+', score))
```

```python
# Поиск на обратной цепи
rev_seq = sequence.reverse_complement()
for position, score in pwm.search(rev_seq, threshold=threshold):
    orig_pos = len(sequence) - position - len(sites[0])
    hits.append((orig_pos, '-', score))
```

```python
# Сортировка по позиции
hits.sort(key=lambda x: x[0])
```

```python
print(f"Всего найдено хитов: {len(hits)}")
for hit in hits:
    print(f"Позиция: {hit[0]:7d}, Цепь: {hit[1]}, Скор: {hit[2]:.3f}")
```
